```python
# Q1 Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes
#    deduction as below:Salary(Lakhs) : Tax(%)• Below 5 : 0% • 5-10 : 10% • 10-20 : 20% • aboove 20 : 30%


ctc_lakhs = float(input("Enter your CTC in Lakhs: "))
ctc = ctc_lakhs * 100000

# Deductions
hra = 0.10 * ctc
da = 0.05 * ctc
pf = 0.03 * ctc

salary_after_deduction = ctc - (hra + da + pf)

# Tax slab
if ctc_lakhs < 5:
    tax_rate = 0
elif ctc_lakhs <= 10:
    tax_rate = 0.10
elif ctc_lakhs <= 20:
    tax_rate = 0.20
else:
    tax_rate = 0.30


tax = tax_rate * salary_after_deduction
net_yearly_salary = salary_after_deduction - tax


monthly_salary = net_yearly_salary / 12

print("Your monthly in-hand salary is: ₹", monthly_salary)
```

```python
# Q5 Display Fibonacci series up to 10 terms.
#    Note: The Fibonacci Sequence is a series of numbers. The next number is found by adding up the two numbers before it. 
#    The first two numbers are 0 and 1. For example, 0, 1, 1, 2, 3, 5, 8, 13, 21. The next number in this series above is 13+21 = 34

a = 0
b = 1
N = int(input("Enter the number of terms: "))
print("Fibonacci Series")
print(a,end = " ")
print(b,end = " ")

for i in range(N):
    c = a + b
    a = b
    b = c
    print(c,end = " ")
```

```python
# Q8 Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. 
#   And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to 
#   solve this problem.


def recursive_sum(n, total=0):
    if total > 300 or n == 0:
        return total
    
    if n % 5 == 0:
        return recursive_sum(n - 1, total)
        
    return recursive_sum(n - 1, total + n)



N = int(input("Enter a number N: "))
result = recursive_sum(N)
print("Final Sum =", result)
```

```python
# Q10 Write a program which will find all such numbers which are divisible by 7 but are not a multiple of 5, between 2000 and 3200 (both included). 
#     The numbers obtained should be printed in a comma-separated sequence on a single line.

numbers = []

for i in range(2000, 3201):
    if (i % 7 == 0) and (i % 5 != 0):
        numbers.append(str(i))

print(",".join(numbers))
```

```python
# Q11 Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number.
#     The numbers obtained should be printed in a space-separated sequence on a single line.

numbers = []

for num in range(1000, 3001):
    s = str(num)
    if all(int(d) % 2 == 0 for d in s):  
        numbers.append(s)

print(" ".join(numbers))
```

```python
# Q12 A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps. 
#     The trace of robot movement is shown as the following:UP 5, DOWN 3, LEFT 3, RIGHT 2, !

x = 0
y = 0

while True:
    move = input("Enter move (or ! to stop): ")
    if move == "!":
        break
    
    d, s = move.split()
    s = int(s)

    if d == "UP": y += s
    elif d == "DOWN": y -= s
    elif d == "LEFT": x -= s
    elif d == "RIGHT": x += s

print("Final position:", (x, y))
```

```python
# Q14 Print all the Armstrong numbers in a given range. Range will be provided by the user Armstrong number is a number that is equal to 
#     the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

N = int(input("Enter start of range: "))
E = int(input("Enter end of range: "))

print(f"Armstrong numbers between {N} and {E} are:")

for num in range(N, E + 1):
    sum_of_cubes = 0
    temp = num
    while temp > 0:
        digit = temp % 10
        sum_of_cubes += digit ** 3
        temp //= 10

    if num == sum_of_cubes:
        print(num, end=" ")
   
```

```python
# decimal to binary converter without bin function
number = int(input("Enter Number: "))
list_decimal = []
while number > 0:
    rem = number % 2
    list_decimal.append(rem)
    number = number // 2
for ele in list_decimal[::-1]:
    print(ele,end="")
```

```python
# LCM and HCF of the two number

def HCF(x,y):
    while(y):
        # x=y
        # y=x%y
        x, y = y, x % y
    return x
def LCM(x,y):
    return (x * y) // HCF(x, y)

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
d = HCF(num1,num2)
print("The HCF of the number: ",d)
c = LCM(num1,num2)
print("The LCM of the number: ",c)
```

```python
# LCM and HCF of the two number 

X = int(input("Enter first number: "))
Y = int(input("Enter second number: "))

# LCM of number's
if X > Y:
    greater = X
else:
    greater = Y
while True:
    if (greater%X==0) and (greater%Y==0):
        LCM = greater
        print("The LCM of the numbers is: ",LCM)
        break
    greater += 1

# HCF of number's
if X < Y:
    smaller = X
else:
    smaller = Y
for i in range(2,smaller+1):
    if (X%i==0) and (Y%i==0):
        HCF = i
print("The HCF of the number is: ",HCF)
        
```
